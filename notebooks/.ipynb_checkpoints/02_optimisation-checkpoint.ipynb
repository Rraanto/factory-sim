# Planning optimisation formulations

This notebook compares direct gradient descent in PyTorch with constrained SLSQP from SciPy. Both call the same differentiable twin and therefore optimise the same energy and delivery-failure cost.

```python
from pathlib import Path
import sys
import numpy as np
import torch
from scipy.optimize import minimize

project_root = Path.cwd()
if not (project_root / 'modules').exists():
    project_root = project_root.parent
sys.path.insert(0, str(project_root))
from modules.digital_twin import FactoryDigitalTwin
from utils.scenarios import example_parameters

torch.set_default_dtype(torch.float64)
twin = FactoryDigitalTwin(example_parameters())
shape = (twin.horizon, twin.products)
```

## 1. Penalised formulation with Adam

A sigmoid parameterisation enforces non-negative production. A quadratic penalty discourages schedules that require more line-days than the available production lines.

```python
raw_plan = torch.zeros(shape, requires_grad=True)
optimizer = torch.optim.Adam([raw_plan], lr=0.15)
capacity = twin.parameters.production_capacity * twin.parameters.lines

for _ in range(500):
    optimizer.zero_grad()
    plan = torch.sigmoid(raw_plan) * capacity
    loss = twin.objective(plan, capacity_penalty=1_000.0)
    loss.backward()
    optimizer.step()

adam_plan = (torch.sigmoid(raw_plan) * capacity).detach()
adam_result = twin.simulate(adam_plan)
print('Adam cost:', adam_result.total_cost.item())
print('Minimum remaining capacity:', twin.capacity_constraints(adam_plan).min().item())
print('Success rate:', adam_result.delivery_success_rate.item())
```

## 2. Explicitly constrained formulation with SciPy SLSQP

SLSQP receives gradients from PyTorch. The capacity constraints are expressed exactly as remaining line-days greater than or equal to zero.

```python
def tensor_plan(flat, requires_grad=False):
    return torch.tensor(flat.reshape(shape), dtype=torch.float64, requires_grad=requires_grad)

def objective_and_gradient(flat):
    plan = tensor_plan(flat, requires_grad=True)
    value = twin.objective(plan)
    value.backward()
    return value.item(), plan.grad.detach().numpy().ravel()

def remaining_capacity(flat):
    return twin.capacity_constraints(tensor_plan(flat)).detach().numpy()

def remaining_capacity_jacobian(flat):
    plan = tensor_plan(flat, requires_grad=True)
    jacobian = torch.autograd.functional.jacobian(twin.capacity_constraints, plan)
    return jacobian.detach().numpy().reshape(twin.horizon, -1)

upper_bounds = np.tile((twin.parameters.production_capacity * twin.parameters.lines).numpy(), twin.horizon)
solution = minimize(
    objective_and_gradient,
    x0=np.minimum(adam_plan.numpy().ravel(), upper_bounds),
    jac=True,
    bounds=list(zip(np.zeros_like(upper_bounds), upper_bounds)),
    constraints={'type': 'ineq', 'fun': remaining_capacity, 'jac': remaining_capacity_jacobian},
    method='SLSQP',
    options={'maxiter': 500, 'ftol': 1e-9},
)
slsqp_result = twin.simulate(tensor_plan(solution.x))
print(solution.message)
print('SLSQP cost:', slsqp_result.total_cost.item())
print('Statuses:', slsqp_result.delivery_statuses)
```
